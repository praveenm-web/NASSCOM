```python
import os
import datetime
import random
```

```python
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0), # Celsius
    'Humidity_Sensor': (30.0, 80.0),   # Percentage
    'Pressure_Sensor': (980.0, 1030.0),# hPa
    'Light_Sensor': (100, 1000)        # Lux
}
```

```python
print("\n--- Task 1: Data Generation and Basic Structures ---")
```

```python
raw_sensor_data = []

base_time = datetime.datetime.now()

for i in range(100):
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    timestamp = (base_time + datetime.timedelta(minutes=i)).strftime('%Y-%m-%d %H:%M:%S')
    device_id = f"DEV{i+1:03d}"

    reading = (timestamp, device_id, device_type, location, value)
    raw_sensor_data.append(reading)
```

```python
temperature_readings = []

for reading in raw_sensor_data:
    if reading[2] == 'Temperature_Sensor':
        temp_dict = {
            'timestamp': reading[0],
            'device_id': reading[1],
            'location': reading[3],
            'temperature_celsius': reading[4]
        }
        temperature_readings.append(temp_dict)
```

```python
unique_locations = {reading[3] for reading in raw_sensor_data}

print(unique_locations)
```

```python
device_stats = {}

for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

for reading in raw_sensor_data:
    dtype = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[dtype]

    device_stats[dtype] = (
        min(current_min, value),
        max(current_max, value)
    )

print(device_stats)
```

```python
def get_readings_by_location(readings_list, target_location):
    return [reading for reading in readings_list
            if reading[3] == target_location]
```

```python
def calculate_average_temperature(temp_data_list):

    if not temp_data_list:
        return 0.0

    total = sum(item['temperature_celsius']
                for item in temp_data_list)

    return total / len(temp_data_list)
```

```python
def generate_report_line(reading_dict):

    return (f"[{reading_dict['timestamp']}] "
            f"Device {reading_dict['device_id']} "
            f"at {reading_dict['location']}: "
            f"{reading_dict['temperature_celsius']}°C")
```

# 3.2 Calculate the duration between the first and last timestamp in 'raw_sensor_data'.
#     Convert string timestamps to datetime objects first.
#     Print the duration in days, hours, and minutes.

```python
from datetime import datetime

# Get the first and last raw sensor data entries
first_reading = raw_sensor_data[0]
last_reading = raw_sensor_data[-1]

# Extract timestamps (which are the first element in each tuple)
first_timestamp_str = first_reading[0]
last_timestamp_str = last_reading[0]

# Convert string timestamps to datetime objects
# The format is '%Y-%m-%d %H:%M:%S'
first_datetime = datetime.strptime(first_timestamp_str, '%Y-%m-%d %H:%M:%S')
last_datetime = datetime.strptime(last_timestamp_str, '%Y-%m-%d %H:%M:%S')

# Calculate the duration
duration = last_datetime - first_datetime

# Extract days, hours, and minutes from the timedelta
total_seconds = duration.total_seconds()
days = duration.days
hours = int(total_seconds // 3600 % 24)
minutes = int(total_seconds // 60 % 60)

print(f"Duration between first and last timestamp: {days} days, {hours} hours, {minutes} minutes.")
```
